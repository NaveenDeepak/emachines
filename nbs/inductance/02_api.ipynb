## Overview

Re-exports the full public API for `emachines.inductance`.

Importing from `emachines.inductance` gives access to all public symbols without needing to know which submodule they live in.

## Usage

```python
from emachines.inductance import *
# or selectively:
from emachines.inductance import StatorGeometry, SlotGeometry, airgap_inductance, slot_leakage_inductance
from emachines.inductance import EndTurn, stator_inductances, compare_windings
```

## Test: every public function is re-exported

````python
#| hide
import importlib
import emachines.inductance as inductance
````

````python
#| hide
for sub in ['geometry', 'airgap', 'slot', 'endwinding', 'stator']:
    mod = importlib.import_module(f'emachines.inductance.{sub}')
    missing = [n for n in mod.__all__ if n not in inductance.__all__]
    assert not missing, f"emachines.inductance.{sub} public names not re-exported: {missing}"
    for n in mod.__all__:
        assert getattr(inductance, n) is getattr(mod, n)
````
