## Moved to `emachines.magnetics`

The permeability models describe the material, not the solver, so they now live in **`emachines.magnetics`** (notebook `nbs/magnetics/06_permeability_models.ipynb`), where every module can use them.

This module re-exports them so existing code keeps working:


````python
#| hide
# The old import paths must give the very same objects as the new home
import emachines.magnetics as magnetics
import emachines.mec as mec
import emachines.mec.material as material

for name in material.__all__:
    assert getattr(material, name) is getattr(magnetics, name), f"emachines.mec.material.{name} is not the magnetics object"
for name in ['PermeabilityModel', 'LinearPermeabilityModel', 'SplinePermeabilityModel', 'ShanesudhoffModel']:
    assert getattr(mec, name) is getattr(magnetics, name), f"emachines.mec.{name} is not the magnetics object"
print("✓ emachines.mec and emachines.mec.material re-export the magnetics permeability models")
````
