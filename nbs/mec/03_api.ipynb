## Overview

Re-exports the full public API for `emachines.mec`.

Importing from `emachines.mec` gives access to all public symbols without needing to know which submodule they live in.

## Usage

```python
from emachines.mec import *
# or selectively:
from emachines.mec import PermeabilityModel
from emachines.mec import BranchType
from emachines.mec import MECSolution
```

````python
#| default_exp mec.__init__
#| export
from emachines.mec.material import PermeabilityModel, LinearPermeabilityModel, SplinePermeabilityModel, ShanesudhoffModel
from emachines.mec.mec import BranchType, MEC
from emachines.mec.result import MECSolution

__all__ = ['PermeabilityModel', 'LinearPermeabilityModel', 'SplinePermeabilityModel', 'ShanesudhoffModel', 'BranchType', 'MEC', 'MECSolution']
````
