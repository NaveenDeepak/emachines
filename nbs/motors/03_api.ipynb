## Overview

Re-exports the full public API for `emachines.motors`.

Importing from `emachines.motors` gives access to all public symbols without needing to know which submodule they live in.

## Usage

```python
from emachines.motors import *
# or selectively:
from emachines.motors import GeometricParams
from emachines.motors import PMSMParams
```

````python
#| default_exp motors.__init__
#| export
from emachines.motors.dc_motor import (
    GeometricParams,
    MaterialProperties,
    OperatingConditions,
    LumpedParams,
    calculate_carter_coefficient,
    calculate_magnetic_circuit,
    calculate_armature_resistance,
    calculate_motor_constants,
    calculate_rotor_inertia,
    calculate_lumped_parameters,
    calculate_steady_state_performance,
    calculate_losses_and_efficiency,
    plot_efficiency_map,
)
from emachines.motors.pmsm import PMSMParams, back_emf, torque, dq_currents, SPM

__all__ = [
    'GeometricParams',
    'MaterialProperties',
    'OperatingConditions',
    'LumpedParams',
    'calculate_carter_coefficient',
    'calculate_magnetic_circuit',
    'calculate_armature_resistance',
    'calculate_motor_constants',
    'calculate_rotor_inertia',
    'calculate_lumped_parameters',
    'calculate_steady_state_performance',
    'calculate_losses_and_efficiency',
    'plot_efficiency_map',
    'PMSMParams',
    'back_emf',
    'torque',
    'dq_currents',
    'SPM',
]
````

````python
#| hide
import importlib
import emachines.motors as motors

for sub in ['dc_motor', 'pmsm']:
    mod = importlib.import_module(f'emachines.motors.{sub}')
    missing = [n for n in mod.__all__ if n not in motors.__all__]
    assert not missing, f"emachines.motors.{sub} public names not re-exported: {missing}"
    for n in mod.__all__:
        assert getattr(motors, n) is getattr(mod, n), f"{n} differs between package and {sub}"
assert all(hasattr(motors, n) for n in motors.__all__), "__all__ lists a name that is not defined"
assert len(motors.__all__) == len(set(motors.__all__)), "duplicate names in __all__"
print(f"✓ emachines.motors re-exports all {len(motors.__all__)} public names")
````
