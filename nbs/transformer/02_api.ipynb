## Overview

Re-exports the full public API for `emachines.transformer`.

Importing from `emachines.transformer` gives access to all public symbols without needing to know which submodule they live in.

## Usage

```python
from emachines.transformer import *
# or selectively:
from emachines.transformer import TransformerCore, no_load_current
```

````python
#| default_exp transformer.__init__
#| export
from emachines.transformer.single_phase import (
    TransformerCore,
    TransformerWinding,
    CoreMaterial,
    Supply,
    peak_flux_density,
    no_load_current,
    core_loss,
    magnetising_branch,
    winding_resistance,
    rogowski_factor,
    leakage_inductance,
    load_performance,
)

__all__ = [
    'TransformerCore',
    'TransformerWinding',
    'CoreMaterial',
    'Supply',
    'peak_flux_density',
    'no_load_current',
    'core_loss',
    'magnetising_branch',
    'winding_resistance',
    'rogowski_factor',
    'leakage_inductance',
    'load_performance',
]
````

````python
#| hide
import importlib
import emachines.transformer as transformer

for sub in ['single_phase']:
    mod = importlib.import_module(f'emachines.transformer.{sub}')
    missing = [n for n in mod.__all__ if n not in transformer.__all__]
    assert not missing, f"emachines.transformer.{sub} public names not re-exported: {missing}"
    for n in mod.__all__:
        assert getattr(transformer, n) is getattr(mod, n), f"{n} differs between package and {sub}"
assert all(hasattr(transformer, n) for n in transformer.__all__), "__all__ lists a name that is not defined"
assert len(transformer.__all__) == len(set(transformer.__all__)), "duplicate names in __all__"
print(f"✓ emachines.transformer re-exports all {len(transformer.__all__)} public names")
````
