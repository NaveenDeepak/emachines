`emachines` is an open-source Python library of well-documented, tested, and citable implementations of the analytical models used in electric motor design. It is the physics engine behind [emdesignlabs.com](https://emdesignlabs.com).

Every function documents the equation it implements and its bibliographic source.

## Installation

```bash
pip install emachines
```

Requires Python >= 3.10.

## Modules

| Module | Capability |
|--------|------------|
| [Winding](winding/01_factors.ipynb) | Winding factors (kp, kd, kw), MMF distribution, star-of-slots phasor method |
| [Motors](motors/01_pmsm.ipynb) | PMSM dq-frame model, DC motor analytical model |
| [Magnetics](magnetics/01_bh_models.ipynb) | BH curve models, iron loss (Steinmetz, Bertotti), electrical steel & PM databases |
| [MEC](mec/02_solver.ipynb) | Magnetic equivalent circuit Newton-Raphson solver with nonlinear materials |

## Quick Start

````python
from emachines.winding import winding_factor
from emachines.magnetics.iron_loss import bertotti
from emachines.motors.pmsm import PMSMParams, torque

# Winding factor -- 12s/10p FSCW tooth-coil
kw = winding_factor(5, Q=12, P=10, w=1)   # mechanical order: working harmonic ν = p = 5
print(f'kw1 = {kw:.4f}')  # 0.9330

# Iron loss -- Bertotti forward model
loss = bertotti(f=400, B_peak=1.5, k_h=0.02, k_e=1e-4, k_a=1e-3)
print(f'Total loss = {loss["total"]:.2f} W/kg')

# PMSM torque
motor = PMSMParams(p=3, Ld=5e-3, Lq=8e-3, psi_m=0.15, Rs=0.1)
Te = torque(motor, id=-3.0, iq=10.0)
print(f'Torque = {Te:.2f} Nm')
````

````python
#| hide
# The Quick Start above is the README; check it prints what its comments say.
import math
assert math.isclose(kw, 0.9330, abs_tol=1e-4), kw
assert math.isclose(Te, 7.155, abs_tol=1e-3), Te
````

## Design Philosophy

- **Equations first** -- every function leads with the formula it implements
- **Cited** -- every formula references a specific textbook, paper, or standard
- **Notebook-driven** -- source lives in Jupyter notebooks; Python modules are auto-generated via [nbdev](https://nbdev.fast.ai/)
- **Tested** -- validated against published datasets and reference tools
- **Dependency-light** -- only numpy, scipy, pandas, and matplotlib required
