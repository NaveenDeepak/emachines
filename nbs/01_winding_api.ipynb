## Overview

Re-exports the full public API for `emachines.winding`.

Importing from `emachines.winding` gives access to all public symbols without needing to know which submodule they live in.

## Usage

```python
from emachines.winding import *
# or selectively:
from emachines.winding import pitch_factor
from emachines.winding import mmf_distribution
from emachines.winding import get_basic_params
```

````python
#| default_exp winding.__init__
#| export
from emachines.winding.winding_factors import pitch_factor, distribution_factor, winding_factor, winding_factor_spectrum
from emachines.winding.mmf import mmf_distribution, mmf_waveform, mmf_harmonics
from emachines.winding.sos import get_basic_params, build_star_of_slots, assign_phases

__all__ = ['pitch_factor', 'distribution_factor', 'winding_factor', 'winding_factor_spectrum', 'mmf_distribution', 'mmf_waveform', 'mmf_harmonics', 'get_basic_params', 'build_star_of_slots', 'assign_phases']
````
