## Overview

Re-exports the full public API for `emachines.magnetics`.

Importing from `emachines.magnetics` gives access to all public symbols without needing to know which submodule they live in.

## Usage

```python
from emachines.magnetics import *
# or selectively:
from emachines.magnetics import MODEL_NAMES
from emachines.magnetics import linear_region
from emachines.magnetics import SAMPLE_BH
from emachines.magnetics import MAGNET_LIBRARY
```
