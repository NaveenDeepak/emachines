## Overview

Re-exports the public API of `emachines.gears.helical`: helical gear geometry and pair checks, layout (tooth counts, helix angle choice, profile shift), ISO 6336 rating with helix factors, shafts, bearings and hubs, manufacturing data, and the design driver.

All quantities are in millimetres, newtons, N·m and MPa, as in gear standards and drawings. The spur and planetary tools stay in `emachines.gears`.

## Usage

```python
#| eval: false
from emachines.gears.helical import HelicalRequirements, design_helical_pair

req = HelicalRequirements(ratio=3.0, torque_in=20.0, torque_peak=40.0, n_in=3000.0, life_h=2000,
                          shaft_d1=15.0, priority="balanced")     # or "quiet", "low_thrust"
best = design_helical_pair(req)[0]
print(best.label())
best.write_package("helical_package")     # report, gear data tables, parts list, layout SVG, DXF outlines
```

## Test: every public function is re-exported

```python
#| hide
import importlib
import emachines.gears.helical as helical

for sub in ['geometry', 'layout', 'rating', 'components', 'manufacturing', 'design']:
    mod = importlib.import_module(f'emachines.gears.helical.{sub}')
    missing = [n for n in mod.__all__ if n not in helical.__all__]
    assert not missing, f"emachines.gears.helical.{sub} public names not re-exported: {missing}"
    for n in mod.__all__:
        assert getattr(helical, n) is getattr(mod, n)
```
