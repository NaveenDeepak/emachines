## Overview

Re-exports the full public API for `emachines.gears`: involute spur gear geometry, single-stage planetary layout, ISO 6336 rating, component sizing, manufacturing data and the design driver.

All gear quantities are in millimetres, newtons, N·m and MPa, as in gear standards and drawings.

## Usage

```python
#| eval: false
from emachines.gears import PlanetaryRequirements, design_planetary

req = PlanetaryRequirements(ratio=5.0, torque_out=20.0, n_in=3000.0,
                            envelope_d=120.0, envelope_length=40.0, shaft_d=14.0)
best = design_planetary(req)[0]
print(best.label())
best.write_package("planetary_package")      # report, gear data tables, parts list, layout SVG, DXF outlines
```

## Test: every public function is re-exported

```python
#| hide
import importlib
import emachines.gears as gears

for sub in ['involute', 'layout', 'rating', 'components', 'manufacturing', 'design']:
    mod = importlib.import_module(f'emachines.gears.{sub}')
    missing = [n for n in mod.__all__ if n not in gears.__all__]
    assert not missing, f"emachines.gears.{sub} public names not re-exported: {missing}"
    for n in mod.__all__:
        assert getattr(gears, n) is getattr(mod, n)
```
