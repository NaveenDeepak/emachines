## Overview

Re-exports the full public API for `emachines.winding`.

Importing from `emachines.winding` gives access to all public symbols without needing to know which submodule they live in.

## Usage

```python
from emachines.winding import *
# or selectively:
from emachines.winding import pitch_factor
from emachines.winding import mmf_distribution
from emachines.winding import get_basic_params
from emachines.winding import design_windings
```

````python
#| default_exp winding.__init__
#| export
from emachines.winding.winding_factors import pitch_factor, distribution_factor, winding_factor, winding_factor_spectrum
from emachines.winding.mmf import mmf_distribution, mmf_waveform, mmf_harmonics
from emachines.winding.sos import (get_basic_params, build_star_of_slots, assign_phases, check_winding,
                                   build_coil_matrix, winding_factor_sos, check_symmetry,
                                   get_valid_coil_spans, is_valid_combination)
from emachines.winding.design import WindingCandidate, design_windings

__all__ = ['pitch_factor', 'distribution_factor', 'winding_factor', 'winding_factor_spectrum',
           'mmf_distribution', 'mmf_waveform', 'mmf_harmonics',
           'get_basic_params', 'build_star_of_slots', 'assign_phases', 'check_winding',
           'build_coil_matrix', 'winding_factor_sos', 'check_symmetry',
           'get_valid_coil_spans', 'is_valid_combination',
           'WindingCandidate', 'design_windings']
````

## Test: every public function is re-exported

Each winding module lists its public names in `__all__`. This check fails if any of them is missing from `emachines.winding`, so a new function cannot be added to a module without also being added here.

````python
#| hide
import importlib
import emachines.winding as winding

for sub in ['winding_factors', 'mmf', 'sos', 'design']:
    mod = importlib.import_module(f'emachines.winding.{sub}')
    missing = [n for n in mod.__all__ if n not in winding.__all__]
    assert not missing, f"emachines.winding.{sub} public names not re-exported: {missing}"
    for n in mod.__all__:
        assert getattr(winding, n) is getattr(mod, n), f"{n} differs between package and {sub}"
assert all(hasattr(winding, n) for n in winding.__all__), "__all__ lists a name that is not defined"
assert len(winding.__all__) == len(set(winding.__all__)), "duplicate names in __all__"
print(f"✓ emachines.winding re-exports all {len(winding.__all__)} public names")
````
