## Theory: NdFeB Magnets

**Strongest permanent magnets available.**

**Properties:**
- Remanence (Br): ≈1.0–1.5 T at 20°C
- Coercivity (Hcb, Hcj): Depends on grade
- Temperature effects: Br and Hcj decline

**Grades:**
- N-series (80°C): Room temperature
- M-series (100°C): Moderate temperature
- H/SH/UH/EH (120–200°C): High temperature

```python
#| export
@dataclass
class MagnetGrade:
    """
    Room-temperature (20 °C) properties of a permanent magnet grade.

    Units (as stored in MAGNET_LIBRARY):
        br        : T
        hcb, hcj  : kA/m   (MagnetData expects A/m; use MagnetData.from_grade)
        alpha_br  : %/K    temperature coefficient of Br
        alpha_hcj : %/K    temperature coefficient of Hcj
    """
    grade:     str
    br:        float
    hcb:       float
    hcj:       float
    alpha_br:  float
    alpha_hcj: float
```

```python
#| export
@dataclass
class MagnetData:
    """
    Temperature-corrected demagnetization curves.

    Units: Br in T, Hcj in **A/m**, alpha_br and alpha_hcj in %/K, temperatures in °C.
    To build one from a MAGNET_LIBRARY grade (stored in kA/m), use ``from_grade``.
    """
    grade:         str
    from_standard: str
    temperature_C: float
    T_max:         float
    Br:            float
    Hcj:           float
    alpha_br:      float
    alpha_hcj:     float
    
    H:     np.ndarray = field(default_factory=lambda: np.array([]))
    J:     np.ndarray = field(default_factory=lambda: np.array([]))
    B:     np.ndarray = field(default_factory=lambda: np.array([]))
    J_20C: np.ndarray = field(default_factory=lambda: np.array([]))
    B_20C: np.ndarray = field(default_factory=lambda: np.array([]))
    
    @classmethod
    def from_grade(cls, grade: MagnetGrade, temperature_C: float, T_max: float,
                   from_standard: str = "MAGNET_LIBRARY") -> "MagnetData":
        """Create MagnetData from a MagnetGrade, converting Hcj from kA/m to A/m."""
        return cls(grade=grade.grade, from_standard=from_standard, temperature_C=temperature_C,
                   T_max=T_max, Br=grade.br, Hcj=grade.hcj * 1e3,
                   alpha_br=grade.alpha_br, alpha_hcj=grade.alpha_hcj)

    def calculate_JH(self) -> None:
        """
        Compute temperature-corrected J-H and B-H curves.

        Raises ValueError if Hcj looks like kA/m instead of A/m, and warns if
        temperature_C is above T_max.
        """
        if self.Hcj < 1e4:
            raise ValueError(
                f"Hcj = {self.Hcj} looks like kA/m; MagnetData expects A/m "
                f"(e.g. {self.Hcj * 1e3:.0f}). Use MagnetData.from_grade() for MAGNET_LIBRARY grades.")
        if self.temperature_C > self.T_max:
            warnings.warn(
                f"{self.grade}: temperature {self.temperature_C} °C is above the maximum operating "
                f"temperature {self.T_max} °C; the curve is extrapolated beyond the grade's rating.",
                stacklevel=2)
        delta_T  = self.temperature_C - 20
        beta_br  = self.alpha_br  * delta_T * 1e-2
        beta_hcj = self.alpha_hcj * delta_T * 1e-2
        Br_temp  = self.Br  * (1 + beta_br)
        Hcj_temp = self.Hcj * (1 + beta_hcj)
        self.H = np.linspace(-self.Hcj, 0, 100)
        self.J, self.B = self._jh_data(self.H, Br_temp, -Hcj_temp)
        self.J_20C, self.B_20C = self._jh_data(self.H, self.Br, -self.Hcj)
    
    def _jh_data(self, H: np.ndarray, Br_temp: float, Hcj_temp: float, k1: float = -6e-5) -> tuple[np.ndarray, np.ndarray]:
        """Exponential J-H model for NdFeB."""
        u0 = 4 * np.pi * 1e-7
        ur = 1.05
        k2 = np.log(Br_temp + (ur - 1) * u0 * Hcj_temp) / k1 - Hcj_temp
        J2 = np.exp(k1 * (k2 + H))
        J_cal = Br_temp + u0 * (ur - 1) * H - J2
        B_cal = Br_temp + u0 * ur * H - J2
        return J_cal, B_cal
```

Verify magnet grades and temperature correction.

```python
# Test MAGNET_LIBRARY
assert "N42" in MAGNET_LIBRARY
assert len(MAGNET_LIBRARY) == 25
n42 = MAGNET_LIBRARY["N42"]
assert n42.br == 1.305
print(f"✓ MAGNET_LIBRARY: {len(MAGNET_LIBRARY)} grades")

# Test MagnetData
magnet = MagnetData(
    grade="N42",
    from_standard="Arnold Magnetics",
    temperature_C=80,
    T_max=80,
    Br=n42.br,
    Hcj=n42.hcj * 1e3,
    alpha_br=n42.alpha_br,
    alpha_hcj=n42.alpha_hcj,
)
magnet.calculate_JH()
assert len(magnet.H) == 100
assert len(magnet.B) == 100
print(f"✓ MagnetData.calculate_JH(): {len(magnet.H)} points")
print("\nAll tests passed!")

# ── Units: library grades are kA/m, MagnetData wants A/m ──────────────
import math, warnings
m1 = MagnetData.from_grade(n42, temperature_C=20, T_max=80)
m1.calculate_JH()
assert m1.Hcj == n42.hcj * 1e3
assert math.isclose(m1.J[-1], n42.br, rel_tol=1e-9), m1.J[-1]       # J(H=0) = Br
assert abs(m1.J[0]) < 1e-9                                          # J(H=-Hcj) = 0
bad = MagnetData(grade="N42", from_standard="lib", temperature_C=20, T_max=80, Br=n42.br,
                 Hcj=n42.hcj, alpha_br=n42.alpha_br, alpha_hcj=n42.alpha_hcj)   # kA/m by mistake
try:
    bad.calculate_JH()
    assert False, "Hcj in kA/m should raise"
except ValueError as e:
    assert "looks like kA/m" in str(e)
print("✓ MagnetData.from_grade and kA/m guard")

# ── T_max warning ─────────────────────────────────────────────────────
with warnings.catch_warnings(record=True) as caught:
    warnings.simplefilter("always")
    MagnetData.from_grade(n42, temperature_C=120, T_max=80).calculate_JH()
    MagnetData.from_grade(n42, temperature_C=60, T_max=80).calculate_JH()
assert len(caught) == 1 and "above the maximum operating temperature" in str(caught[0].message)
print("✓ MagnetData warns above T_max")
```
